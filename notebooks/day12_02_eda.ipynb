**Task 3 - Derive date parts from InvoiceDate, PODate**

```python
## Convert Date Columns
# Purchases Sample
purchases_sample["InvoiceDate"] = pd.to_datetime(purchases_sample["InvoiceDate"])
purchases_sample["PODate"] = pd.to_datetime(purchases_sample["PODate"])

# Vendor Invoice
vendor_invoice["InvoiceDate"] = pd.to_datetime(vendor_invoice["InvoiceDate"])
vendor_invoice["PODate"] = pd.to_datetime(vendor_invoice["PODate"])

print("Date columns converted successfully.")
```

```python
## Derive Date Parts
# Purchases Sample
purchases_sample["Invoice_Year"] = purchases_sample["InvoiceDate"].dt.year
purchases_sample["Invoice_Month"] = purchases_sample["InvoiceDate"].dt.month
purchases_sample["Invoice_Day"] = purchases_sample["InvoiceDate"].dt.day

purchases_sample["PO_Year"] = purchases_sample["PODate"].dt.year
purchases_sample["PO_Month"] = purchases_sample["PODate"].dt.month
purchases_sample["PO_Day"] = purchases_sample["PODate"].dt.day

# Vendor Invoice
vendor_invoice["Invoice_Year"] = vendor_invoice["InvoiceDate"].dt.year
vendor_invoice["Invoice_Month"] = vendor_invoice["InvoiceDate"].dt.month
vendor_invoice["Invoice_Day"] = vendor_invoice["InvoiceDate"].dt.day

vendor_invoice["PO_Year"] = vendor_invoice["PODate"].dt.year
vendor_invoice["PO_Month"] = vendor_invoice["PODate"].dt.month
vendor_invoice["PO_Day"] = vendor_invoice["PODate"].dt.day

print("Date parts derived successfully.")
```
