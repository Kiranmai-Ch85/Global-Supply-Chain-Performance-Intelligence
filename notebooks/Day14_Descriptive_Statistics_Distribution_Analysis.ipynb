# Document Sampling Caveats

### Sampling Caveats

1. Some datasets are sampled subsets rather than complete enterprise records, so the analysis may not fully represent all business transactions.

2. Extreme outliers are present in Quantity, Price, and Volume columns. These values may represent genuine bulk purchases or premium products, but they can influence summary statistics.

3. Different datasets contain different row counts and business scopes, so direct comparison across all datasets should be done carefully.

4. Some datasets are aggregated (for example, Procurement KPI and Vendor Sales Summary), while others contain transaction-level records. Their results should be interpreted according to their level of detail.

5. Findings from this analysis provide useful business insights, but decisions should ideally be validated using complete production data before implementation.
