**purchase_prices**

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Admin\Documents\Enterprise Procurement & Vendor Performance Analytics\data\raw\purchase_prices.csv")

print(df.shape)
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
# Remove rows where Description, Size or Volume are missing
df = df.dropna(subset=["Description", "Size", "Volume"])

# Convert Volume to numeric
# Any non-numeric value (like "Unknown") becomes NaN
df["Volume"] = pd.to_numeric(df["Volume"], errors="coerce")

# Remove rows where Volume became NaN
df = df.dropna(subset=["Volume"])

# Convert Volume to integer
df["Volume"] = df["Volume"].astype(int)

print(df.shape)
```

```python
df.isnull().sum()
```

```python
df = pd.read_csv(
    r"C:\Users\Admin\Documents\Enterprise Procurement & Vendor Performance Analytics\data\python data cleaning\purchase_prices_cleaned.csv"
)

print(df.shape)

# Check for new lines in Description
print(df[df["Description"].astype(str).str.contains("\n|\r", regex=True, na=False)])

# Check datatypes
print(df.dtypes)

# Save again
df.to_csv(
    r"C:\Users\Admin\Documents\Enterprise Procurement & Vendor Performance Analytics\data\python data cleaning\purchase_prices_cleaned_new.csv",
    index=False,
    encoding="utf-8",
    lineterminator="\n"
)
```

**Procurement KPI Analysis Dataset**

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Admin\Documents\Enterprise Procurement & Vendor Performance Analytics\data\raw\Procurement KPI Analysis Dataset.csv")

print(df.shape)
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
# Fill missing delivery dates
df["Delivery_Date"] = df["Delivery_Date"].fillna("Not Delivered")

# Convert Order_Date to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")

# Convert only valid Delivery_Date values
mask = df["Delivery_Date"] != "Not Delivered"
df.loc[mask, "Delivery_Date"] = pd.to_datetime(
    df.loc[mask, "Delivery_Date"],
    errors="coerce"
)

print(df.info())
```

```python
df.isnull().sum()
```

```python
df.to_csv(
    r"C:\Users\Admin\Documents\Enterprise Procurement & Vendor Performance Analytics\data\day 11 python cleaning\Procurement_KPI_Analysis_Dataset_cleaned.csv",
    index=False
)

print("Procurement_KPI_Analysis_Dataset_cleaned.csv saved successfully")
```
