```python
## Day 13 - Task 1 - Outlier Analysis (IQR Method)

def detect_outliers_iqr(df, column, dataset_name):

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower) | (df[column] > upper)]

    print("="*80)
    print(dataset_name)
    print("Column :", column)
    print("Lower Limit :", round(lower,2))
    print("Upper Limit :", round(upper,2))
    print("Outliers :", len(outliers))
```

```python
detect_outliers_iqr(purchase_prices,"Price","Purchase Prices")
detect_outliers_iqr(purchase_prices,"PurchasePrice","Purchase Prices")
detect_outliers_iqr(purchase_prices,"Volume","Purchase Prices")

detect_outliers_iqr(purchases_sample,"Quantity","Purchases Sample")
detect_outliers_iqr(purchases_sample,"PurchasePrice","Purchases Sample")

detect_outliers_iqr(sales_sample,"SalesQuantity","Sales Sample")
detect_outliers_iqr(sales_sample,"SalesPrice","Sales Sample")
detect_outliers_iqr(sales_sample,"Volume","Sales Sample")

detect_outliers_iqr(begin_inventory,"Price","Begin Inventory")
detect_outliers_iqr(end_inventory,"Price","End Inventory")

detect_outliers_iqr(procurement,"Quantity","Procurement KPI")
detect_outliers_iqr(procurement,"Unit_Price","Procurement KPI")

detect_outliers_iqr(purchase_orders,"ordered_qty","Purchase Orders")
detect_outliers_iqr(purchase_orders,"received_qty","Purchase Orders")

detect_outliers_iqr(vendor_invoice,"Quantity","Vendor Invoice")
```

**Task 3 - Dataset-Grounded Insights**

**Insight 1 – Purchase Prices Dataset**

The Purchase Prices dataset contains a considerable number of outliers in Price, PurchasePrice, and Volume, indicating that a small group of products are significantly more expensive or have larger packaging sizes than the majority of products.

**Insight 2 – Purchases Sample Dataset**

The Purchases Sample dataset shows several outliers in Quantity and PurchasePrice, suggesting that some procurement transactions involve bulk purchasing or premium-priced items compared to regular purchase transactions.

**Insight 3 – Sales Sample Dataset**

The Sales Sample dataset contains high outliers in SalesQuantity, SalesPrice, and Volume, indicating that a limited number of products contribute disproportionately to overall sales quantity and revenue.

**Insight 4 – Begin Inventory Dataset**

The Begin Inventory dataset shows noticeable outliers in Price, indicating that the inventory consists of both regular-priced and premium-priced products at the beginning of the inventory period.

**Insight 5 – End Inventory Dataset**

Similar to the Begin Inventory dataset, the End Inventory dataset also contains price outliers, suggesting that high-value products remain in inventory at the end of the reporting period.

**Insight 6 – Procurement KPI Dataset**

The Procurement KPI dataset contains only a few outliers in Quantity and no significant outliers in Unit_Price, indicating that procurement activities are generally consistent and follow standard purchasing practices.

**Insight 7 – Purchase Orders Dataset**

The Purchase Orders dataset contains outliers in both ordered_qty and received_qty, indicating that some purchase orders involve exceptionally large quantities compared to normal procurement transactions.

**Insight 8 – Vendor Invoice Dataset**

The Vendor Invoice dataset shows a limited number of quantity outliers, suggesting that only a small number of vendor invoices represent unusually large deliveries while most invoices fall within the expected operational range.
